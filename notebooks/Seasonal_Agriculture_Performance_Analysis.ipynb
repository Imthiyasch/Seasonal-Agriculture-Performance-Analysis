This notebook analyzes seasonal differences in agricultural performance using the supplied dataset.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
df = pd.read_csv("../data/seasonal_agriculture_performance_dataset.csv")
print(df.shape)
df.head()
```

## 1. Data quality

```python
print("Missing cells:", int(df.isna().sum().sum()))
print("Duplicate rows:", int(df.duplicated().sum()))
print(df.isna().sum()[df.isna().sum()>0])
```

## 2. Seasonal performance

```python
season_summary=df.groupby("Season").agg(mean_yield="Yield_Tonnes_Ha",mean_profit="Profit_INR",mean_rainfall="Rainfall_mm",mean_temperature="Avg_Temperature_C",mean_humidity="Humidity_pct",mean_water_used="Water_Used_m3",mean_water_efficiency="Water_Efficiency_t_per_1000m3").round(2)
season_summary
```

## 3. Irrigation and resource use

```python
irrigation_summary=df.groupby("Irrigation_Method").agg(mean_yield="Yield_Tonnes_Ha",mean_profit="Profit_INR",mean_water_used="Water_Used_m3",mean_water_efficiency="Water_Efficiency_t_per_1000m3").sort_values("mean_profit",ascending=False).round(2)
irrigation_summary
```

## 4. Crop ተ season

```python
crop_season=df.pivot_table(index="Crop",columns="Season",values="Yield_Tonnes_Ha",aggfunc="mean").round(2)
crop_season
```

## 5. Regional consistency

```python
state_season=df.pivot_table(index="State",columns="Season",values="Yield_Tonnes_Ha",aggfunc="mean").round(2)
state_season
```

## 6. Correlation analysis

```python
profit_corr=df.select_dtypes(include=np.number).corr(numeric_only=True)["Profit_INR"].sort_values(ascending=False)
profit_corr.head(10)
```
